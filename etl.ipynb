```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\sonim\OneDrive\Documents\codes\cl-4\ETL\sales_dataset(ETL - Data Visualization).csv")
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date']) 
```

```python
df = df.drop_duplicates()                       
```

```python
df = df.dropna()                                 # Remove null values
```

```python
sales_summary = df.groupby('Product Category')['Total Amount'].sum()
```

```python
print("Processed Data:")
print(sales_summary)
```

```python
plt.bar(sales_summary.index, sales_summary.values)
plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
daily_sales = df.groupby('Date')['Total Amount'].sum()

plt.plot(daily_sales.index, daily_sales.values)
plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.pie(
    sales_summary.values,
    labels=sales_summary.index,
    autopct='%1.1f%%'
)
plt.title("Sales Distribution by Category")
plt.show()
```
